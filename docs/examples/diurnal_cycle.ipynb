# Diurnal cycle

Fit the diurnal cycle of air temperature.

```python
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr

import xclimatology  # registers the .climatology accessor
```

## Load the data

We use 6-hourly NCEP reanalysis air temperature near the surface over North America for
2013–2014, from the xarray tutorial datasets.

```python
air = xr.tutorial.load_dataset("air_temperature")["air"]
air
```

## Remove the seasonal cycle

Subtract the seasonal climatology, the mean, annual and semiannual harmonics, and linear
trend, to obtain the seasonal anomaly.

```python
seasonal = dict(n_harmonics=2, period=1, period_units="year", detrend_degree=1)

seasonal_anomaly = air - air.climatology(**seasonal)
```

## Fit the diurnal cycle

`seasonal_anomaly.climatology(**diurnal)` fits, independently at each grid point,

$$
y(t) = \mu + a \cos(\omega t) + b \sin(\omega t) + \varepsilon(t),
\qquad \omega = \frac{2\pi}{T},
$$

where $t$ is time in days, $T = 1$ day, $\mu$ is the mean value, and $\varepsilon$ is the
residual. `n_harmonics=1` keeps the diurnal harmonic alone, `period` and `period_units`
set $T$, and `detrend_degree=None` omits the trend, which the seasonal climatology already
removed. The mean, the 0th harmonic, is always fit. Phases are measured from midnight of the time coordinate, which is 00 UTC for
these data.

```python
diurnal = dict(n_harmonics=1, period=1, period_units="day", detrend_degree=None)

diurnal_clim = seasonal_anomaly.climatology(**diurnal)
diurnal_clim
```

At a single grid point, over two weeks of July:

```python
point = dict(lat=40, lon=250)
july = slice("2013-07-01", "2013-07-14")

fig, ax = plt.subplots(figsize=(10, 3.5))
seasonal_anomaly.sel(**point).sel(time=july).plot(ax=ax, marker=".", lw=0.5, label="seasonal anomaly")
diurnal_clim.sel(**point).sel(time=july).plot(ax=ax, marker=".", lw=1.5, label="diurnal climatology")
ax.set(title="40°N, 110°W", ylabel="K")
ax.legend(loc="upper left");
```

The fit uses one amplitude for both years, so it underestimates the larger diurnal
range of summer.

## Amplitude and phase

`seasonal_anomaly.climatology.coefficients(**diurnal)` returns the fitted coefficients with
the amplitude and phase of the diurnal harmonic,

$$
a \cos(\omega t) + b \sin(\omega t) = A \cos(\omega t - \phi),
\qquad A = \sqrt{a^2 + b^2}, \qquad \phi = \mathrm{atan2}(b, a).
$$

The diurnal harmonic $A \cos(\omega t - \phi)$ peaks when its argument is a multiple of
$2\pi$, $\omega t - \phi = 2\pi n$. The phase from `atan2` lies between $-\pi$ and $\pi$,
so taking it modulo $2\pi$ gives the first peak after 00 UTC,

$$
t_1 = \frac{\phi \bmod 2\pi}{\omega},
$$

in days. This is the timing of the peak of the fitted cosine, which repeats every day,
and not the time of the largest observed value. In local solar time, the peak falls at
approximately

$$
\left(24\,t_1 + \frac{\lambda}{15^\circ}\right) \bmod 24
$$

hours, where $\lambda$ is longitude east.

```python
coefs = seasonal_anomaly.climatology.coefficients(**diurnal).sel(harmonic=1)
hour_utc = (coefs["phase"] % (2 * np.pi)) / (2 * np.pi) * 24
hour = (hour_utc + air["lon"] / 15) % 24

fig, axes = plt.subplots(ncols=2, figsize=(11, 4), constrained_layout=True)
coefs["amplitude"].plot.contourf(ax=axes[0], levels=21, cbar_kwargs={"label": "K"})
axes[0].set_title("Diurnal amplitude")
hour.plot.contourf(
    ax=axes[1], cmap="twilight", levels=np.arange(0, 25),
    cbar_kwargs={"label": "local solar hour", "ticks": np.arange(0, 25, 3)},
)
axes[1].set_title("Hour of the diurnal peak");
```

Over land the diurnal harmonic peaks in mid-afternoon, and its amplitude is largest over
the deserts of the southwestern United States and northwestern Mexico.
